```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset=pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv' ,header=1)
# header = 1  =>  skip first row, use second row as column names
```

```python
dataset.head()
```

```python
dataset.info()
```

```python
dataset.isnull().sum()
```

```python
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.loc[:122, 'Region']=0
dataset.loc[122:, 'Region']=1
df=dataset
```

```python
df[['Region']] = df[['Region']].astype(int)
```

```python
df.head()
```

```python
df[df.isnull().any(axis=1)]
```

```python
# Removing null values
df = df.dropna().reset_index(drop=True)
# reset_index => Resets the index to make it continuous again
# drop=True => Prevents old index from becoming a new column
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df = df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
# Fix spaces in column names
df.columns = df.columns.str.strip()
df.columns
```

#### Changes the required columns as integer data type

```python
df[['month','day','year','Temperature','RH','Ws']]=df[['month','day','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

#### Changing the other columns to float data datatype(Better approach)

```python
strings = [features for features in df.columns if df[features].dtypes=='str']
```

```python
for i in strings:
  if i!= 'Classes':
    df[i]=df[i].astype(float)
```

```python
df.info()
```

```python
df[['Classes']] = df[['Classes']].astype(object)
```

```python
df.info()
```

```python
# Saving the cleaned dataset
df.to_csv('Algerian_forest_fires_cleaned_dataset', index=False)
```
