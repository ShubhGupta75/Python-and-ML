```python
import pandas as pd
messages=pd.read_csv('SMSSpamCollection.csv', sep='\t', names=['label', 'message'])
```

```python
messages
```

```python
# Data cleaning and Preprocessing
import nltk
import re
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
corpus=[]
for i in range(0,len(messages)):
  review=re.sub('[^a-zA-Z]',' ', messages['message'][i])  # Replace all non-alphabetic characters with spaces
  review=review.lower() # Convert the text to lowercase
  review=review.split() # Split the text into individual words (tokens)
  review=[ps.stem(word) for word in review if word not in stopwords.words('english')] # Remove stopwords and apply stemming
  review=" ".join(review)
  corpus.append(review)
```

```python
# Output Features
import pandas as pd

y=pd.get_dummies(messages['label'])
y=y.iloc[:,0].values
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.20)
```

```python
# Bag of Words
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=2500, ngram_range=(1,2))
```

```python
len(X_train),len(y_train)
```

```python
## independent features
X_train=cv.fit_transform(X_train).toarray()
X_test=cv.transform(X_test).toarray()
```

```python
# Train ML model
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_detect_model=MultinomialNB().fit(X_train,y_train)
```

```python
y_pred=spam_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

### Using TF-IDF

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.20)
```

```python
# TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
tv=TfidfVectorizer(max_features=2500,ngram_range=(1,2))
X_train=tv.fit_transform(X_train).toarray()
X_test=tv.transform(X_test).toarray()
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_tfidf_model=MultinomialNB().fit(X_train,y_train)
```

```python
y_pred=spam_tfidf_model.predict(X_test)
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
```

```python
print(classification_report(y_test,y_pred))
```
