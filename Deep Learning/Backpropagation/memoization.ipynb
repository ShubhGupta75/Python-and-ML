```python
# Improving the time complexity of fibonacci series calculation
def fib(n,d):
  if n in d:
    return d[n]
  else:
    d[n] = fib(n-1,d) + fib(n-2,d)
    return d[n]
```

```python
d={0:1, 1:1}
fib(50,d)
# The dictionary stores previously computed Fibonacci values (memoization). When the same subproblem is encountered again, its stored result is returned instead of recalculating it.
```
